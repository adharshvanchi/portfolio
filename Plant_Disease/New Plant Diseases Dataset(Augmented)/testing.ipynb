```python
import pandas as pd
import cv2 as cv
import os
```

Step 1: Creating a dataframe with im_names and labels


```python
# Creating the Master DF
train_dir = '/Users/adharshv/Documents/Career/portfolio/Plant_Disease/New Plant Diseases Dataset(Augmented)/train/'
dir_ls = os.listdir(train_dir)
df = pd.DataFrame({'im_name':[],'plant_type':[],'status':[]})

for dir in dir_ls:
    temp_df = pd.DataFrame()
    dir_name = dir.split('_',1)
    plant_name = dir_name[0]
    status = dir_name[1].replace('_','')
    files = os.listdir(train_dir+dir)
    temp_df['im_name'] = files
    temp_df['plant_type'] = plant_name
    temp_df['status'] = status
    df = pd.concat([df,temp_df],axis=0,ignore_index=True)
    # print(plant_name)
    # print(status)
    # print(files)
    
df
```
